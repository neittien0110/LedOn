```python
from pynq import Overlay
import time

# 1. Nạp thiết kế vào FPGA
ol = Overlay("../LedOn/LedOn.bit") 

# 2. Truy cập vào GPIO (Tên này phải khớp với tên trong Vivado)
leds = ol.axi_gpio_0

# Cấu hình 4 bit đầu tiên thành Output (giá trị 0)
# Số 0 nghĩa là tất cả 4 bit đều thành Output
leds.mmio.write(0x04, 0x0)

# 3. Thử bật 2 LED sáng, 2 LED tối (Số 12 = 1100 nhị phân)
print("Đang bật LED...")
leds.write(0, 5) 

# 4. Kiểm tra xem LED trên board có sáng không?
time.sleep(2)
leds.write(0, 15) # Sáng hết cả 4 LED
print("Đã bật toàn bộ LED!")
```

```python
from pynq import MMIO
# 0x41200000 là địa chỉ mặc định của GPIO đầu tiên trong Vivado
# Nếu địa chỉ của bạn khác, hãy sửa lại theo file .hwh
mmio = MMIO(0x41200000, 0x10) 

# Ghi giá trị 15 (1111) vào thanh ghi dữ liệu (offset 0)
mmio.write(0, 11) 

# Ghi giá trị 0 vào thanh ghi hướng (offset 4) để ép nó làm Output
mmio.write(4, 0)
```
